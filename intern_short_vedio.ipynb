```python
import pandas as pd
import yt_dlp

df_link = pd.read_excel("/content/video_metadata.xlsx")

ydl_opts = {
    "quiet": True,
    "no_warnings": True,
    "skip_download": True
}

with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    for index, row in df_link.iterrows():
        try:
            info = ydl.extract_info(row["video_url"], download=False)
            df_link.loc[index, "duration_sec"] = info.get("duration")
            print(row["video_id"], info.get("duration"))

        except Exception as e:
            print("Error:", row["video_id"], e)

df_link.to_csv("/content/video_metadata_updated.csv", index=False)
```

```python
from youtube_transcript_api import YouTubeTranscriptApi

video_id = "rKpltaOMFdc"

api = YouTubeTranscriptApi()

transcript = api.fetch(video_id)

transcript_data = []

for item in transcript:
    transcript_data.append({
        "start_time": item.start,
        "duration": item.duration,
        "text": item.text
    })

print(transcript_data[:10])
```

```python
import pandas as pd

transcript_df = pd.DataFrame(transcript_data)

transcript_df.head(10)
```

```python
transcript_df.to_csv(
    "/content/V001_transcript.csv",
    index=False
)
```

```python
transcript_df["end_time"] = (
    transcript_df["start_time"] + transcript_df["duration"]
)

transcript_df["video_id"] = "V001"

transcript_df = transcript_df[
    ["video_id", "start_time", "end_time", "text"]
]

transcript_df.head()
```

```python
from difflib import SequenceMatcher

def combine_transcript(df):
    df = df.sort_values("start_time").reset_index(drop=True)

    combined = []

    for _, row in df.iterrows():
        text = str(row["text"]).strip()

        if not combined:
            combined.append({
                "start_time": row["start_time"],
                "end_time": row["end_time"],
                "text": text
            })
            continue

        previous = combined[-1]

        if row["start_time"] < previous["end_time"]:
            previous_words = previous["text"].split()
            current_words = text.split()

            max_overlap = min(len(previous_words), len(current_words))

            overlap = 0

            for n in range(1, max_overlap + 1):
                if previous_words[-n:] == current_words[:n]:
                    overlap = n

            remaining_text = " ".join(current_words[overlap:])

            previous["text"] += " " + remaining_text
            previous["end_time"] = max(
                previous["end_time"],
                row["end_time"]
            )

        else:
            combined.append({
                "start_time": row["start_time"],
                "end_time": row["end_time"],
                "text": text
            })

    return pd.DataFrame(combined)


clean_transcript = combine_transcript(transcript_df)

clean_transcript.head(20)
```
